```python
micropolis_network = IntegratedNetwork(name = 'Micropolis')
```

```python
MAIN_DIR = Path('../..')
network_dir= 'infrarisk/data/networks/micropolis'
water_folder = MAIN_DIR/f'{network_dir}/water/med'
power_folder = MAIN_DIR/f'{network_dir}/power/low'
transp_folder = MAIN_DIR/f'{network_dir}/transportation/high'

# load all infrastructure networks
micropolis_network.load_networks(water_folder, 
                                 power_folder, 
                                 transp_folder, 
                                 power_sim_type = '1ph', 
                                 water_sim_type = "PDA")

scenario_location = MAIN_DIR/network_dir/"scenarios"
```

```python
micropolis_network.generate_integrated_graph()
```

# Point event: Explosion

```python
explosion = hazard.RadialDisruption(point_of_occurrence=(6410, 4130), 
                                    radius_of_impact= 100, 
                                    intensity = "extreme",
                                    name = "Micropolis explosion")
```

```python
explosion.set_affected_components(micropolis_network.integrated_graph)
```

```python

explosion.generate_disruption_file(location = scenario_location)
```

# Track-based event: Micropolis flood

```python
micropolis_folder = Path('../../infrarisk/data/networks/micropolis')
micropolis_streams = gpd.read_file(micropolis_folder/'gis/streams_v1.shp', encoding="utf-8")

flood = hazard.TrackDisruption(hazard_tracks = micropolis_streams[micropolis_streams.index == 1], 
                               buffer_of_impact = 100, 
                               time_of_occurrence=6000, 
                               intensity = "moderate", 
                               name = "Micropolis flood")
```

```python
flood.set_affected_components(micropolis_network.integrated_graph, plot_components=True)
```

```python
flood.generate_disruption_file(location = Path(scenario_location)/'flood1')
```

# Track-based event based on custom tracks: Micropolis Hurricane

```python
hurricane = hazard.TrackDisruption(hazard_tracks = None, 
                               buffer_of_impact=100, 
                               time_of_occurrence=6000,
                               intensity = "high",
                               name = "Micropolis hurricane")

micropolis_map_extents = micropolis_network.get_map_extends()
hurricane_track = hurricane.generate_random_track(micropolis_map_extents, shape = "spline")
hurricane.set_hazard_tracks_from_linestring(hurricane_track)
```

```python

hurricane.set_affected_components(micropolis_network.integrated_graph, plot_components=True)
```

## Track events can also be created for a specific part of the network

```python
tornado = hazard.TrackDisruption(hazard_tracks = None, 
                               buffer_of_impact=50, 
                               time_of_occurrence=6000,
                               intensity = "moderate",
                               name = "Micropolis tornado")

micropolis_tornado_extents = [(2000, 3000), (3000, 5000)]
tornado_track = tornado.generate_random_track(micropolis_tornado_extents, shape = "line")
tornado.set_hazard_tracks_from_linestring(tornado_track)
```

```python
tornado.set_affected_components(micropolis_network.integrated_graph, plot_components=True)
```
