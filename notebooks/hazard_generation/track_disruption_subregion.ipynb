# Track disruption subregion (Tornadoes)
Track disruption occurs along a track. The tracks can either be randomly generated using spline functions. Alternatively, for disasters, such as floods, where the course of the flood is known, the track can be sepcified. The intensity of the event is highest along the centerline of the track, and dissipates to zero at the edges of the buffer area specified by the user. Here we generate random tracks and then generate disruptions along the track. For events, such as tornadoes, the track may be shorter. Such random events can be generated by specifying a subregion.

### Load the micropolis network

```python
micropolis_network = IntegratedNetwork(name = 'Micropolis')
```

```python
MAIN_DIR = Path('../..')
network_dir= 'infrarisk/data/networks/micropolis'
water_folder = MAIN_DIR/f'{network_dir}/water/med'
power_folder = MAIN_DIR/f'{network_dir}/power/low'
transp_folder = MAIN_DIR/f'{network_dir}/transportation/high'

# load all infrastructure networks
micropolis_network.load_networks(water_folder, 
                                 power_folder, 
                                 transp_folder, 
                                 power_sim_type = '1ph', 
                                 water_sim_type = "PDA")
micropolis_network.generate_integrated_graph()
```

### Generate a subregional track disruption

```python
tornado = TrackDisruption(hazard_tracks = None, 
                               buffer_of_impact=50, 
                               time_of_occurrence=6000,
                               intensity = "moderate",
                               name = "Micropolis tornado")

micropolis_tornado_extents = [(2000, 3000), (3000, 5000)]
tornado_track = tornado.generate_random_track(micropolis_tornado_extents, shape = "line")
tornado.set_hazard_tracks_from_linestring(tornado_track)
```

```python
tornado.set_affected_components(micropolis_network.integrated_graph, plot_components=True)
```

### Write the disruption file to local directory

```python
scenario_location = MAIN_DIR/network_dir/"scenarios"
tornado.generate_disruption_file(location = scenario_location)
```
