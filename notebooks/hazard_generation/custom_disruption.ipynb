# Custom disruption (user defined)
The available disruption scenarios in the package do not consider the specific characteristics of the hazards and the fragility of the infrastructure components. For more accurate analysis, an external hazard- and fragility analysis model can be used to generate direct disruptions in the networks and then use it as input for network level simulations using InfraRisk. This notebook shows how to use the custom disruption scenario to generate disruptions in the network.

```python
%load_ext autoreload
%autoreload 2

from pathlib import Path
from infrarisk.src.physical.integrated_network import IntegratedNetwork
from infrarisk.src.hazards.custom import CustomDisruption
```

### Load the micropolis network

```python
micropolis_network = IntegratedNetwork(name = 'Micropolis')
```

```python
MAIN_DIR = Path('../..')
network_dir= 'infrarisk/data/networks/micropolis'
water_folder = MAIN_DIR/f'{network_dir}/water/med'
power_folder = MAIN_DIR/f'{network_dir}/power/low'
transp_folder = MAIN_DIR/f'{network_dir}/transportation/high'

# load all infrastructure networks
micropolis_network.load_networks(water_folder, 
                                 power_folder, 
                                 transp_folder, 
                                 power_sim_type = '1ph', 
                                 water_sim_type = "PDA")
micropolis_network.generate_integrated_graph()
```

### Generate a custom disruption
The specific names of the failed components are specified the user.

```python
list_of_failed_compons = ['W_PMA3', 'T_L20', 'P_L3', 'T_L126', 'T_L106', 'W_PMA728']
openquake_disruptions  = CustomDisruption(affected_components = list_of_failed_compons, 
                                       name = "Openquake disruption",
                                       time_of_occurrence=6000)
```

```python
openquake_disruptions.set_affected_components(micropolis_network.integrated_graph)
```

### Write the disruption file to local directory

```python
scenario_location = MAIN_DIR/network_dir/"scenarios"
openquake_disruptions.generate_disruption_file(location = scenario_location)
```
