# Random disruption
Random disruption produces random component failures in the three infrastructure networks. The number of failed components in each network is set by the user

```python
MAIN_DIR = Path('../..')
network_dir= 'infrarisk/data/networks/micropolis'
water_folder = MAIN_DIR/f'{network_dir}/water/med'
power_folder = MAIN_DIR/f'{network_dir}/power/low'
transp_folder = MAIN_DIR/f'{network_dir}/transportation/high'

# load all infrastructure networks
micropolis_network.load_networks(water_folder, 
                                 power_folder, 
                                 transp_folder, 
                                 power_sim_type = '1ph', 
                                 water_sim_type = "PDA")
micropolis_network.generate_integrated_graph()
```

### Generate a radial disruption
The number of failed components in each network is set using an array in the order water, power, and transportation

```python
compon_count_dict = {'water': 5, 'power': 5, 'transpo': 5}
random_disruptions  = RandomDisruption(failure_count=compon_count_dict, 
                                       time_of_occurrence=6000,
                                       name = "Random disruption")
```

```python
fail_compon_dict = {"power": ["L"],
                    "water": ["PMA", "WP"],
                    "transport": ["L"]}

random_disruptions.set_fail_compon_dict(fail_compon_dict = fail_compon_dict)
```

```python
random_disruptions.set_affected_components(micropolis_network.integrated_graph)
```
